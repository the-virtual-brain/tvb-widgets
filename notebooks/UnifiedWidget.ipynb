# Unified Head Widget

Edit a connectivity and see the result in the same place: VIEW shows the connectivity, CONTROL is where you edit it.

**VIEW**
- **Head View:** pick a node from the dropdown and show its incoming or outgoing edges. You can do this for several nodes, each one gets its own color. To see all edges, open `ConnectivityEdges` under Objects in the K3D panel and tick `visible`.
- **Space Time:** the connectivity split into time slices, based on the delays between regions.

**CONTROL**
- **Matrices:** the weights and tract lengths, one quadrant at a time. Click a cell to change its value.
- **Select Nodes:** choose and save the nodes that Edge Operations and the export work on.
- **Edge Operations:** change many edges at once, for example all edges between the selected nodes.
- **Reset Edits:** brings back the original connectivity and clears the selection and the shown edges.
- **Export H5:** saves the edited connectivity in the notebook folder. The edges of the nodes that are not selected are exported as 0.

````python
from tvbwidgets.api import LargeScaleConnectivityWidget
from tvb.datatypes.connectivity import Connectivity

conn = Connectivity().from_file()  
conn.configure()

# color_scheme: name of a matplotlib or colorcet colormap, e.g. "viridis"
wid = LargeScaleConnectivityWidget(conn, color_scheme="turbo") # use wid.conn_edited in following cells
wid.display()
````

````python
conn_edited = wid.conn_edited  # copy of the edited connectivity, rerun this cell after new edits
conn_edited
````
